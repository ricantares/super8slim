```python
import numpy as np
from scipy import stats as st

# data una estrazione di sei numeri, confrontandola con 15 estrazioni precedenti, si ottiene ad esempio che
# il primo estratto si presenta 1 volta,
# il secondo estratto si presenta 1	volta,
# il terzo estratto si presenta 0 volte,
# il quarto estratto si presenta 2 volte,
# il quinto estratto si presenta 1 volta,
# il sesto estratto si presenta 2 volte,
# allora 
#   le ripetizioni totali nell'intervallo sono 1+1+0+2+1+2=7   
#   i numeri con almeno una ripetizione nell'intervallo 6-1=5 (il secondo estratto non ha ripetizioni)

# Inizializza una matrice 5x6 con valori float
matrice = np.array([[15,7,0.47,5,0.33,0.71],
                    [15,5,0.33,3,0.20,0.60],
                    [15,5,0.33,3,0.20,0.60],
                    [15,5,0.33,4,0.27,0.80],
                    [15,9,0.60,6,0.40,0.67]])

# Calcola il minimo, il massimo, la media e la deviazione standard per ogni colonna
minimi = np.min(matrice, axis=0)
massimi = np.max(matrice, axis=0)
medie = np.mean(matrice, axis=0)
mediane = np.median(matrice, axis=0)
mode = st.mode(matrice, axis=0)
deviazioni_standard = np.std(matrice, axis=0)

# Stampa i risultati
print("Minimi:", minimi)
print("Massimi:", massimi)
print("Medie:", medie)
print("Mediane:", mediane)
print("Mode:", mode[0])
print("Deviazioni standard:", deviazioni_standard)

# Matrice di input
matrice_input = matrice

# Conta la frequenza di ripetizioni totali nell'intervallo (colonna 1 della matrice)
j = 1
valori_colonna = matrice_input[:, j]
values, counts = np.unique(valori_colonna, return_counts=True)
print("frequenza di ripetizioni totali nell'intervallo                :", values, counts )

# Conta la frequenza di numeri con almeno una ripetizione nell'intervallo (colonna 3 della matrice)
j = 3
valori_colonna = matrice_input[:, j]
values, counts = np.unique(valori_colonna, return_counts=True)
print("frequenza di numeri con almeno una ripetizione nell'intervallo :", values, counts )
```
