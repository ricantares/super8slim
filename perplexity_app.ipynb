```python
import numpy as np
import pandas as pd

def calcola_ritardi(df):
    # Inverte il DataFrame per processare dalla più vecchia alla più recente
    df_invertito = df[::-1].reset_index(drop=True)
    
    n_estrazioni, n_numeri = df_invertito.shape
    ultima_comparsa = {num: None for num in range(1, 91)}
    ritardi = np.full((n_estrazioni, n_numeri), np.nan)
    
    # Scansiona le estrazioni in ordine cronologico (dalla più vecchia)
    for j in range(n_estrazioni):
        for k in range(n_numeri):
            numero = df_invertito.iat[j, k]
            if ultima_comparsa[numero] is not None:
                ritardi[j, k] = j - ultima_comparsa[numero]
            ultima_comparsa[numero] = j
    
    # Ripristina l'ordine originale (dalla più recente)
    return ritardi[::-1]

def calcola_statistiche(ritardi):
    ritardi_validi = ritardi[~np.isnan(ritardi)]
    return (
        np.mean(ritardi_validi),
        np.median(ritardi_validi),
        np.std(ritardi_validi)
    )

# Esempio di utilizzo con dataset di prova
estrazioni = pd.DataFrame([
    [1, 5, 10, 20, 30, 40],  # Più recente (indice 0)
    [2, 5, 11, 21, 31, 41],
    [1, 6, 12, 22, 32, 42],
    [3, 7, 13, 23, 33, 43],
    [1, 8, 14, 24, 34, 44]   # Più vecchia (indice 4)
])

ritardi = calcola_ritardi(estrazioni)
media, mediana, deviazione = calcola_statistiche(ritardi)

print("Matrice dei ritardi:")
print(ritardi)
print("\nStatistiche:")
print(f"Media: {media:.2f}")
print(f"Mediana: {mediana}")
print(f"Scarto quadratico medio: {deviazione:.2f}")
```
