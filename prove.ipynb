```python
import numpy as np

def count_sequences(array):
  """
  Counts the number of increasing sequences in an array using NumPy, assuming
  that the array is sorted in ascending order and that the array does not
  contain any zero values.

  Args:
    array: A NumPy array.

  Returns:
    The number of increasing sequences in the array.
  """

  count = 0
  for i in range(1, len(array)):
    if i > 1:
      if array[i] == array[i-1] + 1 and array[i-1] != array[i-2] + 1:
        count += 1
    else: 
      if array[i] == array[i-1] + 1:
        count += 1

  return count

# Example usage:
#array = np.array([3, 34, 35, 36, 89, 90])
array = np.array([3, 4, 35, 36, 89, 90])
print(count_sequences(array), np.bincount(array))
```

```python
import numpy as np
import math
from lotteria import Lotteria
from collections import Counter
from scipy import stats


#file_estrazione = "./data/estrazioni_test_90.csv"
file_estrazione = "./data/estrazioni_2023_2006.csv"
ll = Lotteria(file_estrazione)
base = np.empty((0, 6), dtype=int)
base = np.append(base, [np.array([23,40,49,57,68,69])], axis=0)
base = np.append(base, [np.array([11,19,28,34,38,45])], axis=0)
base = np.append(base, [np.array([8,18,41,54,75,79])], axis=0)

#ini = 0
#fin = ll.get_nro_estrazioni_90()+ini
s = ll.get_estrazioni()
mu = np.empty((0, len(s) - ll.get_nro_estrazioni_90()), dtype=int)
for i in range(len(s) - ll.get_nro_estrazioni_90()):
    c = ll.freq_singolo_estratto(s[i], i+1)
    #print(i, s[i], c)
    mu = np.append(mu, np.mean(c))
#print(mu)
print("mu: ", len(mu), np.mean(mu), np.std(mu))
cmu = Counter(mu)
print("cmu: ", len(cmu), cmu)

valori = np.array(list(cmu.keys()))
pesi = np.array(list(cmu.values()))
print("cmu key:   ", valori)
print("cmu value: ", pesi)
media_pesata = np.average(valori, weights=pesi)
print("media pesata: ", media_pesata)

moda = stats.mode(mu)
mediana = np.median(mu)
print("moda: ", moda, " mediana: ", mediana)

#NOTA: media, mediana e moda sono praticamente uguali, quindi posso assumere una distribuzione gaussiana
#      con il campionamento su tutto lo storico

# Calcola la somma dei quadrati delle differenze tra i valori e la media pesata, ponderate per i pesi
somma_quadrati_differenze = np.sum(pesi * (valori - media_pesata) ** 2)

# Calcola la varianza ponderata
varianza_ponderata = somma_quadrati_differenze / np.sum(pesi)

# Calcola la deviazione standard come radice quadrata della varianza ponderata
deviazione_standard = np.sqrt(varianza_ponderata)

print("Deviazione Standard della Media Pesata:", deviazione_standard)

# il calolo di seguito è preso da chatgpt3.5

# Calcola la media delle medie dei campioni
media_media_campioni = np.mean(mu)

# Calcola la varianza della media delle medie
varianza_media_media_campioni = np.var(mu) / len(mu)

# Calcola la deviazione standard della media delle medie
deviazione_std_media_media_campioni = np.sqrt(varianza_media_media_campioni)

media = np.round(media_media_campioni, 4)
std = np.round(deviazione_std_media_media_campioni, 4)
print("Media delle medie dei campioni:", media)
print("Deviazione standard della media delle medie dei campioni:", std)
print("Intervallo al 95%: [", np.round(media - 2 * std, 2), np.round(media + 2 * std, 2), "]")
```
