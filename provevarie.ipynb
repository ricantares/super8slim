```python
import numpy as np
from itertools import combinations

def conta_presenze_singolo(vettore_riga, matrice):
    """
    Conta quante volte ogni elemento del vettore riga appare nella matrice.
    """
    
    risultati = [0] * len(vettore_riga) 
    for i, elem in enumerate(vettore_riga):
        conteggio = np.sum(np.isin(matrice, elem))
        risultati[i] = conteggio.item() # Converti da numpy.int64 a int standard
    return risultati

# Combinazioni (n elementi = len(vettore_riga), in classe k = numero_elementi_combinazione)
#
def conta_presenze_combinazioni(vettore_riga, matrice, numero_elementi_combinazione):
    """
    Conta quante volte ogni combinazione in classe k degli elementi del vettore riga appare nella matrice.
    """
    # Genera tutte le combinazioni del vettore (ordinate)
    combinazioni = list(combinations(sorted(vettore_riga), numero_elementi_combinazione))
        
    risultati = [0] * len(combinazioni)        
    for i, combinazione in enumerate(combinazioni):
        print("combinazione:", combinazione[0], combinazione[1])  # Debug: stampa la combinazione corrente
        # Maschera per righe contenenti tutti gli elementi della combinazione
        ha_tutti = np.ones(matrice.shape[0], dtype=bool) # ritorna un array [len(matrice)] di booleani tutti True
        for elem in combinazione:
            ha_elem = np.any(matrice == elem, axis=1) # ritorna un array [len(matrice)] di booleani tutti False a meno dell'occorrenza di elem se esiste
            ha_tutti = ha_tutti & ha_elem # AND logico per mantenere solo le righe che contengono tutti gli elementi della combinazione
        risultati[i] = np.sum(ha_tutti).item() # Converti da numpy.int64 a int standard
           
    return risultati

# Esempio di utilizzo
vettore = np.sort(np.random.choice(90, size=6)) # singola estrazione
matrice = np.sort(np.random.choice(range(1, 91), size=(15, 6), replace=False), axis=1) # 15 estrazioni precedenti

print("Vettore riga:", vettore)
print("Matrice generata:\n", matrice)

risultati = conta_presenze_singolo(vettore, matrice)
print("\nRipetizioni di ogni numero nell'intervallo     :", risultati)
print("Ripetizioni totali nell'intervallo               :", sum(risultati))
print("Media ripetizioni nell'intervallo                :", round(sum(risultati) / len(matrice), 2))
print("Numeri con almeno una ripetizione nell'intervallo:", np.count_nonzero(risultati))
print("Media numeri con ripetizioni nell'intervallo     :", round(np.count_nonzero(risultati) / sum(risultati), 2))
print ("Media ripetizioni per numero nell'intervallo     :", round(np.count_nonzero(risultati) / sum(risultati), 2))  # Valore atteso teorico

numero_elementi_combinazione = 2
risultati = conta_presenze_combinazioni(vettore, matrice, numero_elementi_combinazione)
print("\nCombinazioni in classe k =", numero_elementi_combinazione)
print("Numero ricorrenze di ogni combinazione                :", risultati)
print("Totale ricorrenze combinazioni nell'intervallo        :", sum(risultati))
print("Media ricorrenze combinazioni nell'intervallo         :", round(sum(risultati) / len(matrice), 2)) 
print("Combinazioni con almeno una ricorrenza nell'intervallo:", np.count_nonzero(risultati))
if sum(risultati) > 0:
    print("Media ricorrenze per combinazione nell'intervallo:", round(np.count_nonzero(risultati) / sum(risultati), 2))  
```
