```python
import sys
import os

# Navigate from test/ to src/ using relative path
notebook_dir = os.path.dirname(os.path.abspath('.'))
src_path = os.path.join(notebook_dir,  'src')
sys.path.insert(0, src_path)

from common_lib import CommonLib

import numpy as np
from itertools import combinations
import json

##############################################################################

data_file_name = "../data/estrazioni_2026_2006_copy.csv"
clib = CommonLib(data_file_name)

# Popolazione
history_data = clib.load_data(50) # Carica le estrazioni da un file CSV

ini = 20
samples = 15
expected_numbers = history_data[ini-1]  # Prima riga (index 0) è la riga di intestazione, quindi index 1 è la prima estrazione

# Extract 15 rows starting from the second row and create a sorted array without duplicates
extracted_rows = history_data[ini:ini+samples]  # Rows from index 1 to 15 (second row to 16th row)

# Flatten all numbers from extracted rows and remove duplicates, cast to Python ints
extracted_numbers = sorted(set(np.array(extracted_rows).flatten().astype(int)))
extracted_numbers = [int(x) for x in extracted_numbers]

print(f"Extracted {len(extracted_rows)} rows")

# Create array with numbers from 1 to 90 NOT found in extracted data
all_numbers = set(range(1, 91))  # All numbers from 1 to 90
not_found_numbers = sorted(all_numbers - set(extracted_numbers))

print(f"Found {len(extracted_numbers)} unique numbers - NOT found in extracted data: {len(not_found_numbers)}")

# Ensure expected_numbers is a flat list of ints
expected_list = [int(x) for x in np.array(expected_numbers).flatten()]

# Two lists: which expected numbers are in extracted_numbers, and which are in not_found_numbers
contained_in_extracted = [x for x in expected_list if x in extracted_numbers]
contained_in_not_found = [x for x in expected_list if x in not_found_numbers]

print(f"Expected: {expected_numbers} - in extracted: {contained_in_extracted} - in not found: {contained_in_not_found}")
```
